```python
Q1 = df[['Sales', 'Quantity', 'Discount', 'Profit']].quantile(0.25)
Q3 = df[['Sales', 'Quantity', 'Discount', 'Profit']].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bounds:")
print(lower_bound)

print("\nUpper Bounds:")
print(upper_bound)
```

```python
outlier_counts = (
    ((df[['Sales', 'Quantity', 'Discount', 'Profit']] < lower_bound) |
     (df[['Sales', 'Quantity', 'Discount', 'Profit']] > upper_bound))
    .sum()
)

print("Number of potential outliers:")
print(outlier_counts)
```

## Sales Distribution

A histogram is used to understand the distribution of sales values and
identify whether sales are concentrated around lower values or spread
across a wider range.

```python
plt.figure(figsize=(10, 6))

plt.hist(df['Sales'], bins=30)

plt.title('Distribution of Sales')
plt.xlabel('Sales')
plt.ylabel('Frequency')

plt.show()
```

## Profit Distribution

A histogram is used to examine the distribution of profit and understand
the frequency of profitable and loss-making transactions.

```python
plt.figure(figsize=(10, 6))

plt.hist(df['Profit'], bins=30)

plt.title('Distribution of Profit')
plt.xlabel('Profit')
plt.ylabel('Frequency')

plt.show()
```

## Sales by Product Category

This analysis compares total sales across different product categories
to identify which categories contribute the most to overall revenue.

```python
category_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)

category_sales
```

```python
plt.figure(figsize=(8, 5))

category_sales.plot(kind='bar')

plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)

plt.show()
```

## Profit by Product Category

This analysis compares total profit across product categories to determine
which categories contribute most to business profitability.

```python
category_profit = df.groupby('Category')['Profit'].sum().sort_values(ascending=False)

category_profit
```

```python
plt.figure(figsize=(8, 5))

category_profit.plot(kind='bar')

plt.title('Total Profit by Category')
plt.xlabel('Category')
plt.ylabel('Total Profit')
plt.xticks(rotation=0)

plt.show()
```

## Sales by Region

This analysis compares total sales across regions to identify differences
in regional sales performance.

```python
region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)

region_sales
```

```python
plt.figure(figsize=(8, 5))

region_sales.plot(kind='bar')

plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)

plt.show()
```

## Sales by Sub-Category

This analysis examines total sales across individual product sub-categories
to identify the specific product groups generating the highest revenue.

```python
subcategory_sales = (
    df.groupby('Sub-Category')['Sales']
    .sum()
    .sort_values(ascending=False)
)

subcategory_sales
```

```python
plt.figure(figsize=(12, 6))

subcategory_sales.plot(kind='bar')

plt.title('Total Sales by Sub-Category')
plt.xlabel('Sub-Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)

plt.show()
```

```python
subcategory_profit = (
    df.groupby('Sub-Category')['Profit']
    .sum()
    .sort_values(ascending=False)
)

subcategory_profit
```

```python
plt.figure(figsize=(12, 6))

subcategory_profit.plot(kind='bar')

plt.title('Total Profit by Sub-Category')
plt.xlabel('Sub-Category')
plt.ylabel('Total Profit')
plt.xticks(rotation=45)

plt.show()
```

```python
subcategory_summary = (
    df.groupby('Sub-Category')
    .agg({
        'Sales': 'sum',
        'Profit': 'sum',
        'Quantity': 'sum'
    })
    .sort_values('Sales', ascending=False)
)

subcategory_summary
```

```python
segment_sales = (
    df.groupby('Segment')['Sales']
    .sum()
    .sort_values(ascending=False)
)

segment_sales
```

```python
plt.figure(figsize=(8, 5))

segment_sales.plot(kind='bar')

plt.title('Total Sales by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)

plt.show()
```

```python
segment_profit = (
    df.groupby('Segment')['Profit']
    .sum()
    .sort_values(ascending=False)
)

segment_profit
```

```python
plt.figure(figsize=(8, 5))

segment_profit.plot(kind='bar')

plt.title('Total Profit by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Total Profit')
plt.xticks(rotation=0)

plt.show()
```

```python
df['Year'] = df['Order Date'].dt.year

yearly_sales = df.groupby('Year')['Sales'].sum()

yearly_sales
```

```python
plt.figure(figsize=(8, 5))

yearly_sales.plot(kind='line', marker='o')

plt.title('Yearly Sales Trend')
plt.xlabel('Year')
plt.ylabel('Total Sales')
plt.grid(True)

plt.show()
```

## Correlation Heatmap

A correlation heatmap shows the strength and direction of relationships
between numerical variables in the dataset.

```python
plt.figure(figsize=(8, 6))

correlation = df[
    ['Sales', 'Quantity', 'Discount', 'Profit', 'Shipping Days']
].corr()

sns.heatmap(
    correlation,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Correlation Heatmap')

plt.show()
```
